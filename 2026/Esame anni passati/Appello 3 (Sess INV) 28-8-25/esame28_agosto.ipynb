## Esercizio 1

Supponiamo che i seguenti valori rappresentino il numero di ore di sonno dormite ogni notte nelle ultime **due settimane**:  
**6.5, 5.0, 7.8, 8.2, 4.9, 6.0, 7.0, 7.5, 6.8, 5.5, 8.0, 6.3, 7.2, 5.8**

1. Inserisci questi valori in un array NumPy

2. Crea un nuovo array che assegna un **punteggio di qualità del sonno** secondo le seguenti regole:
   - `1` se il sonno è **inferiore a 6 ore**
   - `2` se è **tra 6 ore (incluse) e meno di 8 ore**
   - `3` se è **8 ore o più**

3. Calcola la **percentuale di notti** per ciascuna fascia di qualità del sonno:
   - Percentuale di notti con punteggio **1** (meno di 6 ore)
   - Percentuale di notti con punteggio **2** (tra 6 e meno di 8 ore)
   - Percentuale di notti con punteggio **3** (8 ore o più)


#### LEGGENDA

- **np.array([...])** -> Crea un array NumPy

- **array < valore** -> Confronto vettoriale, restituisce un array di booleani

- **array[mask] = valore** -> Assegna un valore a tutti gli elementi selezionati dalla maschera

- **np.sum(mask)** -> Conta i True in una maschera booleana

- **conteggio / len(array) * 100** -> Percentuale

````python
import numpy as np

# 1. Creo l'array
ore_sonno = np.array([6.5, 5.0, 7.8, 8.2, 4.9, 6.0, 7.0, 7.5, 6.8, 5.5, 8.0, 6.3, 7.2, 5.8])
print ("PUNTO 1: ")
print ("Ore di sonno:", ore_sonno.tolist())


# 2. Creo l'array dei punteggi (inizialmente tutto 2, poi modifico)
punteggi = np.full (14, 2, dtype = int)             # 2 = fascia media
punteggi[ore_sonno < 6] = 1                         # Meno di 6 ore
punteggi[ore_sonno >= 8] = 3                        # 8 ore o più

print ("\nPUNTO 2: ")
print("Punteggi della qualità del sonno:", punteggi.tolist())


# 3. Calcolo le percentuali per fascia
notti_1 = np.sum(punteggi == 1)
notti_2 = np.sum(punteggi == 2)
notti_3 = np.sum(punteggi == 3)

perc_1 = (notti_1 / len(ore_sonno)) * 100
perc_2 = (notti_2 / len(ore_sonno)) * 100
perc_3 = (notti_3 / len(ore_sonno)) * 100

print("\nPUNTO 3:")
print(f"Percentuale notti con punteggio 1 (meno di 6 ore): {perc_1:.1f}%")
print(f"Percentuale notti con punteggio 2 (tra 6 e meno di 8 ore): {perc_2:.1f}%")
print(f"Percentuale notti con punteggio 3 (8 ore o più): {perc_3:.1f}%")
````

## Esercizio 2
Un’urna contiene 3 palline rosse e 2 blu. A ogni estrazione si rimette la pallina nell’urna.
Quindi ad esempio la propabilità di estarre una pallina rossa è 3/5.
1. Scrivi una funzione che simula N estrazioni e rappresentale con un array NumPy (1 = rosso, 0 = blu).  Esempio: estrazioni_palline(3) = array([0,0,1])
2. Calcola la proporzione cumulativa di palline rosse dopo 10, 20, 30, …, 100 estrazioni. Siignifica ad esempio che per N estrazioni, contiamo quante volte è uscita una pallina rossa e dividiamo per 10.
3. Fai un grafico dell’andamento della proporzione cumulativa in funzione del numero di estrazioni.



#### LEGGENDA

- **np.random.choice([1,0], size=N, p=[3/5,2/5])** -> Simula N estrazioni con reinserimento (1=rosso, 0=blu)

- **array[:step]** -> Selezione dei primi `step` elementi

- **.mean()** -> Calcola la media (proporzione di 1)

- **plt.plot(step_list, proporzioni, marker='o')** -> Grafico con punti

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Funzione che simula N estrazioni
def estrazioni_palline(N):
    return np.random.choice([1, 0], size = N, p = [3/5, 2/5])

np.random.seed(0)           # Per riproducibilità

print ("PUNTO 1:")
print ("Esempio estrazioni_palline (3):", estrazioni_palline(3))


# 2. Proporzione cumulativa dopo 10, 20, ..., 100 estrazioni
np.random.seed(0)
Nmax = 100
estrazioni_totali = estrazioni_palline(Nmax)

step_list = list(range(10, Nmax + 1, 10))
proporzioni = []

for step in step_list:
    prime_step = estrazioni_totali[:step]
    prop = prime_step.mean()
    proporzioni.append(prop)

print ("\nPUNTO 2 :")
for step, prop in zip(step_list, proporzioni):
    print(f"Dopo {step} estrazioni, proporzione di rosse: {prop:.3f}")



# 3. Grafico
plt.plot(step_list, proporzioni, marker='o')
plt.xlabel('Numero di estrazioni')
plt.ylabel('Proporzione cumulativa di palline rosse')
plt.title('Andamento della proporzione cumulativa (con reinserimento)')
plt.grid(True)
plt.show()
````

## Esercizio 3

1. Genera una matrice NumPy di dimensioni 8 x 8, con valori interi casuali compresi tra -20 e 20 (inclusi).

2. Per ogni riga, seleziona il numero **più vicino a 10**.

#### LEGGENDA

- **np.random.randint(min, max+1, size)** -> Genera interi casuali (estremo superiore escluso, quindi +1)

- **np.abs(array - target)** -> Differenza assoluta tra ogni elemento e il target

- **np.argmin(array)** -> Indice del valore minimo

- **for i in range(n):** -> Ciclo sulle righe

````python
import numpy as np

# 1. Genero la matrice
np.random.seed(0)
matrice = np.random.randint (-20, 21, (8, 8))
print ("PUNTO 1 - Matrice generata (solo prime 2 righe):")
print (matrice[:2])     # stampo solo le prime 2 righe per vedere che funziona


# 2. Per ogni riga, numero più vicino a 10
vicini = []
for i in range(8):
    riga = matrice[i]
    diff = np.abs(riga - 10)                # Distanze assolute dal 10
    indice_min = np.argmin (diff)           # indice del più vicino
    valore = riga[indice_min]
    vicini.append(valore)

vicini_puliti = [int(v) for v in vicini]   # Converto a interi per stampa pulita       
print ("\n PUNTO 2 - Valore più vicino a 10 per ogni riga:")
print (vicini_puliti)
````

## Esercizio 4  

Utilizza il dataset `movies.csv` (contenente colonne come `title`, `budget`, `revenue`, `vote_average`, `vote_count`).  

1. Elimina le righe con i valori mancanti. e **Crea una nuova colonna** `roi` (return on investment), definita come:  

   $$
   \texttt{roi} = \frac{\texttt{revenue} - \texttt{budget}}{\texttt{budget}}
   $$  

   *(Indica di quanto il film ha guadagnato rispetto al costo di produzione).*  

2. Trova i **5 film con ROI più alto** e stampa `title` e `roi`.  

3. **Filtra i film** che hanno:  
   - `budget` > 50 milioni  
   - `vote_average` > 7  

4. **Crea un grafico a barre** con i **10 film con ROI più alto**.  

5. **Crea uno scatter plot** con:  
   - asse x = `budget`  
   - asse y = `revenue`  
   - colore dei punti = `vote_average`  


#### LEGGENDA

- **pd.read_csv('movies.csv')** -> Carica il dataset

- **df.dropna()** -> Elimina le righe con valori mancanti

- **df['roi'] = (df['revenue'] - df['budget']) / df['budget']** -> Calcola il return on investment

- **df.nlargest(n, 'colonna')** -> Restituisce le n righe con i valori più alti nella colonna

- **df[(df['budget'] > 50e6) & (df['vote_average'] > 7)]** -> Filtro con due condizioni

- **plt.bar(x, y)** -> Grafico a barre verticali

- **plt.xticks(rotation=45, ha='right')** -> Ruota le etichette dell'asse x per leggibilità

- **plt.scatter(x, y, c=colore)** -> Grafico a punti, con colore basato su una terza colonna

- **plt.colorbar(label='...')** -> Aggiunge la barra dei colori

- **plt.tight_layout()** -> Regola i margini per evitare tagli

````python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Carico il datset, elimino i NaN e creo i ROI
df = pd.read_csv('movies.csv')
print ("PUNTO 1: ")
print ("- Dimensioni originali:", df.shape)

# Elimino le righe con qualsiasi NaN
df = df.dropna()
print ("- Dimensioni dopo il dropna:", df.shape)

# Escludo i film con budget <= 0 (altrimenti ROI infinito)
df = df[df['budget'] > 0]
print ("- Dopo esclusione budget <=0:", df.shape)

# Creo la colonna ROI
df ['ROI'] = (df ['revenue'] - df ['budget']) / df ['budget']
print ("- Prime 3 righe con le colonne rilevanti:")
print (df[['original_title', 'budget', 'revenue', 'ROI']].head(3))


# 2. 5 Film con ROI più alto
top_5 = df.nlargest (5, 'ROI')
print ("\nPUNTO 2 - TOP 5 ROI:")
for _, row in top_5.iterrows():
    print (f"{row['original_title']} : {row['ROI']:.2f}")


# 3. Filtro budget > 50M e vote_average > 7
filtro = df[(df['budget'] > 50_000_000) & (df['vote_average'] > 7)]
print (f"\nPUNTO 3 - Film con budget > 50M e voto > 7: {len(filtro)}")
if len(filtro) > 0:
    print (filtro[['original_title', 'budget', 'vote_average']].head(3))
else:
    print ("Nessun film soddisfa i criteri.")


# 4. Grafico a barre dei top 10 ROI
top_10 = df.nlargest(10, 'ROI')
plt.figure (figsize = (10, 6))
plt.bar (top_10 ['original_title'], top_10 ['ROI'])
plt.xticks (rotation = 45, ha = 'right')
plt.xlabel ('Titolo')
plt.ylabel ('ROI')
plt.title ('Top 10 film per ROI')
plt.tight_layout ()
plt.show


# 5. Scatter plot budget vs revenue, colore = vote_average
plt.figure(figsize=(10,6))
sc = plt.scatter(df['budget'], df['revenue'], c=df['vote_average'], alpha=0.6)
plt.colorbar(sc, label='Voto medio')
plt.xlabel('Budget (dollari)')
plt.ylabel('Revenue (dollari)')
plt.title('Budget vs Revenue (colore = voto medio)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

## Esercizio 5  

Utilizza il dataset `movies.csv`.  

1. Quante **righe e colonne** ha il dataset? Successivamente elimina la colonna `overview` dal DataFrame.  
2. Trova il **regista** (`director`) che ha diretto il maggior numero di film.  
3. Calcola il **vote_average medio** per ciascun `director` e ordina i primi 10.  
4. Crea un **boxplot** dei valori di `vote_average` raggruppati per `director` (solo i 5 registi con più film).  
5. Crea un **violin plot** della distribuzione dei `vote_average` per genere (`genres`).  


#### LEGGENDA

- **df.shape** -> Numero di righe e colonne

- **df.drop(columns=['col'], inplace=True)** -> Elimina una colonna

- **df['col'].value_counts().idxmax()** -> Valore più frequente

- **df.groupby('col')['voto'].mean().sort_values(ascending=False).head(10)** -> Top 10 medie

- **df[df['col'].isin(lista)]** -> Filtra per valori in una lista

- **sns.boxplot(x='col', y='voto', data=df)** -> Boxplot

- **df['col'].str.split('|').explode()** -> Separa stringhe con | in righe multiple

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carico il dataset e pulizia dei NaN
df = pd.read_csv ('movies.csv')         # Carico il database
df = df.dropna()                        # Pulisco i NaN
df = df[df['budget'] > 0]               # Escludo film con budget < 0


# 1. Dimensioni ed eliminazione colonna overview
df = df.drop(columns = ['overview'])
print ("\nPUNTO 1: ")
print (f"- Righe: {df.shape[0]}, Colonne: {df.shape[1]}")
print ("- Colonne dopo l'eliminazione di overview:", df.shape[1])


# 2. Regista con più film
conteggio_registi = df['director'].value_counts()
top_director = conteggio_registi.idxmax()
n_film = conteggio_registi.max()
print ("\nPUNTO 2:")
print (f"Regista con più film: {top_director} ({n_film} film)")


# 3. Media voto per regista
media_registi = df.groupby('director')['vote_average'].mean().sort_values(ascending = False).head (10)
print ("\nPUNTO 3 - Top 10 registi per media voto: ")
print (media_registi)


# 4. Boxplot per i primi 5 registi con più film
top5_director = conteggio_registi.head(5).index.tolist()
df_top5 = df[df['director'].isin(top5_director)]

print ("\nPUNTO 4 - Boxplot per i primi 5 registi con più film:")
plt.figure (figsize = (10, 6))
sns.boxplot (x = 'director', y = 'vote_average', data = df_top5)
plt.xticks (rotation = 45)
plt.xlabel ('Regista')
plt.ylabel ('Voto medio')
plt.title('Distribuzione voti per i 5 registi con piu film')
plt.show ()


# 5. Violin plot per genere
genres_expanded = df['genres'].str.split('|').explode()     # Separo i generi con ('|')
df_genres = df.loc[genres_expanded.index].copy()
df_genres['genres'] = genres_expanded

print ("\nPUNTO 5 - Violin plot della distribuzione voti per genere:")
plt.figure (figsize = (12, 6))
sns.violinplot (x = 'genres', y = 'vote_average', data = df_genres)
plt.xticks (rotation = 45)
plt.xlabel ('Genere')
plt.ylabel ('Voto medio')
plt.title ('Distribuzione dei voti per genere')
plt.show ()
````

## Esercizio 6  

Hai a disposizione il dataset di coppie `(x, y)`  qui sotto riportato.

1. Usa `scipy.optimize.curve_fit` per stimare i parametri della retta del tipo `y = a * x + b`.  
2. Plotta i punti originali e la retta ottenuta.  
3. Calcola **MAE** e **RMSE** tra i valori reali `y` e quelli stimati dal modello.  


#### LEGGENDA

- **curve_fit(modello, x, y)** -> Restituisce parametri ottimali (popt) e matrice di covarianza (pcov)

- **modello_lineare(x, a, b)** -> Restituisce a*x + b

- **y_pred = a * x + b** -> Valori stimati

- **mean_absolute_error(y, y_pred)** -> MAE

- **np.sqrt(mean_squared_error(y, y_pred))** -> RMSE

- **plt.scatter(x, y)** -> Grafico a punti

- **plt.plot(x, y_pred, 'r-')** -> Retta stimata

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Dati forniti 
x = np.array([
  0.0, 0.3448, 0.6897, 1.0345, 1.3793, 1.7241,
  2.0690, 2.4138, 2.7586, 3.1034, 3.4483, 3.7931,
  4.1379, 4.4828, 4.8276, 5.1724, 5.5172, 5.8621,
  6.2069, 6.5517, 6.8966, 7.2414, 7.5862, 7.9310,
  8.2759, 8.6207, 8.9655, 9.3103, 9.6552, 10.0
])

y = np.array([
   3.99,  3.33,  8.71,  9.83, 10.69, 14.59,
  11.72, 14.87, 16.06, 13.93, 19.61, 19.83,
  18.40, 20.59, 21.45, 26.51, 24.57, 24.94,
  25.55, 27.78, 28.71, 30.87, 34.14, 34.65,
  35.01, 37.94, 38.53, 41.56, 42.28, 42.65
])

# 1. Stima dei parametri
def modello_lineare (x, a, b):
    return a * x + b

popt, pcov = curve_fit(modello_lineare, x, y)
a, b = popt
print ("PUNTO 1 - Parametri stimati")
print(f"- a = {a:.4f} \n- b = {b:.4f}")


# 2. Plot punti e retta
y_pred = modello_lineare (x, a, b)

print ("\nPUNTO 2 - Plot dei punti e della retta:")
plt.figure(figsize=(8,5))
plt.scatter(x, y, label='Dati osservati')
plt.plot(x, y_pred, 'r-', label=f'Retta: y = {a:.2f}x + {b:.2f}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Regressione lineare')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


# 3. MAE e RMSE
mae = mean_absolute_error(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
print("\nPUNTO 3")
print(f"- MAE = {mae:.4f}")
print(f"- RMSE = {rmse:.4f}")

````
